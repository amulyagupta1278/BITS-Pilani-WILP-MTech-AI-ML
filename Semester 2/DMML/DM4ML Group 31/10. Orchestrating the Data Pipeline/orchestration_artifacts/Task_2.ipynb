Creates dummy raw datasets for Telco and Generic churn.

```python
import os, pandas as pd
os.makedirs('data/raw/telco', exist_ok=True)
os.makedirs('data/raw/generic', exist_ok=True)
print('Folders ready.')
```

```python
# Dummy Telco dataset
df_telco = pd.DataFrame({
    'customerID': ['0001','0002','0003'],
    'gender': ['Male','Female','Female'],
    'tenure': [1,34,12],
    'MonthlyCharges': [29.85,85.7,56.95],
    'TotalCharges': [29.85,3320.75,682.6],
    'Churn': ['No','No','Yes']
})
telco_path = 'data/raw/telco/telco_sample.csv'
df_telco.to_csv(telco_path, index=False)
print('Telco saved at', telco_path)
```

```python
# Dummy Generic dataset
df_generic = pd.DataFrame({
    'CustomerId':[1,2,3],
    'CreditScore':[600,720,580],
    'Geography':['France','Spain','Germany'],
    'Age':[40,30,50],
    'Balance':[60000,120000,0],
    'Exited':[0,1,1]
})
gen_path = 'data/raw/generic/generic_churn_sample.csv'
df_generic.to_csv(gen_path, index=False)
print('Generic saved at', gen_path)
```
