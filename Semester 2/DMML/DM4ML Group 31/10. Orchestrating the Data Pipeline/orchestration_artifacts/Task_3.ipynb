# Task 3 — Validation & Cleaning (Papermill/Prefect-safe)
Validates raw CSVs and writes cleaned copies under `data/validated/`.

```python
import os, pandas as pd
os.makedirs('data/validated', exist_ok=True)
os.makedirs('reports', exist_ok=True)

telco_raw = 'data/raw/telco/telco_sample.csv'
gen_raw = 'data/raw/generic/generic_churn_sample.csv'

telco = pd.read_csv(telco_raw)
gen = pd.read_csv(gen_raw)

# Simple validation checks
issues = []
if 'Churn' not in telco.columns:
    issues.append('Telco missing Churn column')
if 'Exited' not in gen.columns:
    issues.append('Generic missing Exited column')

# Fill missing TotalCharges or Balance with median
if 'TotalCharges' in telco.columns:
    telco['TotalCharges'] = pd.to_numeric(telco['TotalCharges'], errors='coerce')
    telco['TotalCharges'] = telco['TotalCharges'].fillna(telco['TotalCharges'].median())
if 'Balance' in gen.columns:
    gen['Balance'] = pd.to_numeric(gen['Balance'], errors='coerce')
    gen['Balance'] = gen['Balance'].fillna(gen['Balance'].median())

telco_out = 'data/validated/telco.csv'
gen_out = 'data/validated/generic.csv'
telco.to_csv(telco_out, index=False)
gen.to_csv(gen_out, index=False)
print('✅ Validated files saved:', telco_out, gen_out)

# Write validation report
with open('reports/validation_report.txt','w') as f:
    f.write('Validation issues:\n')
    for i in issues:
        f.write(f'- {i}\n')
print('📝 Report written to reports/validation_report.txt')
```
