### Task 2: Data Ingestion

```python
import os
from datetime import datetime
import pandas as pd
import logging
```

```python
def ingest_csv(source_path, target_dir):
    try:
        logging.info(f'Starting ingestion from {source_path}')
        df = pd.read_csv(source_path)
        os.makedirs(target_dir, exist_ok=True)
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        target_path = os.path.join(target_dir, f'raw_data_{timestamp}.csv')
        df.to_csv(target_path, index=False)
        logging.info(f'Successfully ingested data and saved to {target_path}')
        return True
    except Exception as e:
        logging.error(f'Error during ingestion from {source_path}: {e}')
        return False
```
