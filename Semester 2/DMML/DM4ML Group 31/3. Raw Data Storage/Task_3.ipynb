````python
def ingest_and_upload(source_path, target_dir):
    df = pd.read_csv(source_path)
    os.makedirs(target_dir, exist_ok=True)
    timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
    target_path = os.path.join(target_dir, f'raw_data_{timestamp}.csv')
    df.to_csv(target_path, index=False)
    print(f"Saved {target_path}")
````

````python
ingest_and_upload('/content/kaggledata/Telco-Customer-Churn.csv', telco_target_dir)
ingest_and_upload('/content/hfdata/customer_churn_dataset.csv', generic_target_dir)
````

## Folder/Bucket Structure Documentation

To ensure efficient management and organization of ingested raw data, the following folder structure has been implemented:
```
DMML_Project/
└── data/
    └── raw/
        ├── telco/
        │    └── raw_data_YYYYMMDD_HHMMSS.csv
        └── generic/
             └── raw_data_YYYYMMDD_HHMMSS.csv
```


### Design Principles

- **Partition by Source:**  
  Data from different sources are stored in separate folders (`telco/` for Telco Customer Churn dataset, `generic/` for the Generic Customer Churn dataset). This separation facilitates easier management and retrieval based on data origin.

- **Partition by Type:**  
  All raw ingested files are stored under the `raw/` directory, distinguishing unprocessed data from any potential processed or transformed datasets in future stages.

- **Partition by Timestamp:**  
  Each ingested file is saved with a timestamp in its filename (`raw_data_YYYYMMDD_HHMMSS.csv`), ensuring version control and traceability of data ingestion over time.

### Benefits

- Enables scalability by cleanly segregating data sources and types.
- Simplifies tracking and auditing of data ingestion activities.
- Supports reproducibility and rollback by maintaining timestamped raw data snapshots.

This structure reflects best practices in data lake management and supports efficient, automated ingestion workflows.
